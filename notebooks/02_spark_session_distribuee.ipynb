```python

from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .master("spark://spark-master:7077")
    .appName("PEDALI-Jalon1")
    .config("spark.driver.host", "pedali-jupyter")
    .config("spark.driver.bindAddress", "0.0.0.0")
    .config("spark.executor.memory", "512m")
    .config("spark.executor.cores", "1")
    .config("spark.cores.max", "2")
    .getOrCreate()
)

sc = spark.sparkContext
sc.setLogLevel("WARN")

print("=== SPARKSESSION PEDALI ===")
print("Spark Version :", spark.version)
print("Master :", sc.master)
print("AppName :", sc.appName)
print("DefaultParallelism :", sc.defaultParallelism)

print("Nombre d'elements :", sc.parallelize(range(1000), 2).count())
```

```python
rdd = sc.parallelize(range(10000), 4)
resultat = rdd.map(lambda x: (x % 5, 1)).reduceByKey(lambda a, b: a + b)
print(resultat.collect())
```
